```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# pairplot. hue is survival.
sns.pairplot(train, hue='Survived', diag_kind='kde', markers=["o", "s"], palette="Set1")
plt.suptitle("Pairplot of Titanic Dataset", y=1.02)
plt.show()
```

```python
# distribution of sibsp
sns.countplot(x='SibSp', hue='Survived', data=train, palette="Set1")
plt.title("Distribution of Siblings/Spouses Aboard vs Survival")
plt.xlabel("Number of Siblings/Spouses Aboard")
plt.ylabel("Count")
plt.legend(title='Survived', loc='upper right', labels=['Did not survive', 'Survived'])
plt.show()
```
