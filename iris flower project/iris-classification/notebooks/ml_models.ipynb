## KNeighbours Classifier
Overview: Classifies a point based on its nearest neighbours in the training set. Imagine you have two classes red and blue. For every new unknown point KNN looks at the closest K points from the training data. The new point is assigned the majority class of those neighbors. 

How it finds the nearest neighbours: Uses a distance metric to measure closeness. The most common is Euclidean distance. You can also use Manhattan distance, cosine similarity, or sutom metrics.

What does K mean: K is the number of neighbors considered. If K = 3 , the algorithm looks at the three nearest points and takes a majority vote. 
* Small K (e.g. 1 or 3): More sensitive to noise (high variance)
* Large K (e.g. 15 or 30): Smoother decision boundaries, but might ignore local structure (more bias)

Key steps --> 
Store the entire training dataset (no training phase—KNN is lazy).
For a new data point:
- Calculate its distance to all training points.
- Select the K closest points.
- Perform majority vote to assign the label.

Done! (No model fitting needed.)


**Strengths**
Simple and intuitive - no complex training. Works well on small, low dimensional datasets. naturally supports Multiclass classification.

**Weaknesses** 
Slow on large datasets (must compute distances on every point). Sensitive to irrelevant or scaled features (features with large values dominate distance) must always standardize or normalize data for KNN. Curse of dimensionality - in high dimensions distances become less meaningful.

**Hyperparameters**
K (number of neighbors): Tune with cross-validation. 
Distance metric: Euclidean or Manhattan tc.
WEights: Uniform (all neighbors count the same) or distance weighted (closer neighbors count more)

````python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5, p=2, metric='minkowski')
knn.fit(x_train, y_train)

knn.score(x_test, y_test)
````

## Logistic Regression

# 📈 Understanding Logistic Regression

Logistic Regression is a **supervised classification algorithm** used to predict a **binary outcome** (e.g., yes/no, 0/1, spam/not spam). Despite the name, it's a **classification model**, not regression.

---

## ✅ 1. Core Idea

Logistic Regression models the **probability** that a given input belongs to a class. It uses a **sigmoid function** to squeeze the output into a range between 0 and 1:

\[
\sigma(z) = \frac{1}{1 + e^{-z}}
\]

Where:
- \( z = w_0 + w_1x_1 + w_2x_2 + \dots + w_nx_n \) is a linear combination of features
- \( \sigma(z) \) is the predicted **probability** of the positive class (e.g., class 1)

---

## 🎯 2. Decision Rule

- If \( \sigma(z) \geq 0.5 \): predict class **1**
- If \( \sigma(z) < 0.5 \): predict class **0**

You can change the threshold (0.5) depending on the problem (e.g., for imbalanced data or medical diagnosis).

---

## 📉 3. Loss Function (Binary Cross-Entropy)

To train the model, Logistic Regression minimizes **log loss**:

\[
\text{Loss} = -\left[y \log(\hat{y}) + (1 - y) \log(1 - \hat{y})\right]
\]

Where:
- \( y \) is the true label (0 or 1)
- \( \hat{y} \) is the predicted probability from the sigmoid function

---

## ⚙️ 4. How It Works

1. Compute weighted sum of input features: \( z = \mathbf{w}^\top \mathbf{x} + b \)
2. Apply sigmoid to get a probability: \( \sigma(z) \)
3. Make a prediction based on a threshold (usually 0.5)
4. During training: update weights using **gradient descent** to minimize loss

---

## 📊 5. Strengths

- **Interpretable**: coefficients show the impact of each feature
- **Efficient**: fast to train, works well on small/medium datasets
- Works well when classes are **linearly separable**
- Output probabilities, not just hard classifications

---

## ⚠️ 6. Limitations

- Assumes **linear relationship** between input features and log-odds
- Can underperform on **non-linear problems** (consider adding interaction terms or switching to tree-based models)
- Sensitive to **outliers** and **multicollinearity**
- Needs **feature scaling** for best performance

---

## Support Vector Machine (SVM)

# 💡 Understanding Support Vector Machines (SVM)

Support Vector Machines (SVMs) are **supervised learning algorithms** used for **classification** and **regression**, though they’re most commonly used for classification.

SVMs are especially powerful for **high-dimensional data** and are effective even with **non-linear** boundaries (via kernels).

---

## ✅ 1. Core Idea

SVM aims to find the **best hyperplane** that separates the classes with the **maximum margin**.

- A **hyperplane** is a line (2D), plane (3D), or higher-dimensional decision boundary.
- The **margin** is the distance between the hyperplane and the closest points from each class.
- SVM tries to **maximize this margin** — making it robust to overfitting.

---

## 🧱 2. Key Concepts

- **Support Vectors**: The data points **closest to the hyperplane**. These define the decision boundary.
- **Margin**: The width between the support vectors of different classes.
- **Hard Margin**: No misclassifications allowed (used when data is perfectly separable).
- **Soft Margin**: Allows some misclassifications (used with real-world, noisy data).

---

## 🧮 3. Mathematical Formulation

For linearly separable data, SVM solves:

\[
\min_{\mathbf{w}, b} \frac{1}{2} ||\mathbf{w}||^2 \quad \text{subject to } y_i(\mathbf{w} \cdot \mathbf{x}_i + b) \geq 1
\]

This is a **convex optimization problem**. For non-linear data, kernels are used.

---

## 🌀 4. Kernel Trick

SVMs can handle non-linearly separable data using the **kernel trick**, which maps data into a higher-dimensional space **without explicitly computing the transformation**.

Common kernels:
- **Linear**: good for simple, linearly separable data
- **Polynomial**
- **RBF (Radial Basis Function)**: most common, handles complex boundaries well
- **Sigmoid**

---

## 🧠 5. How It Works

1. Map data (possibly into higher dimensions using a kernel)
2. Find the hyperplane that **maximizes the margin**
3. Classify new points based on which side of the hyperplane they fall on

---

## 🔧 6. Key Hyperparameters

- `C`: Controls the **trade-off** between maximizing margin and minimizing classification error
  - Small `C` → wider margin, more tolerance for misclassification
  - Large `C` → tighter margin, less tolerance for error
- `kernel`: Choose `'linear'`, `'rbf'`, `'poly'`, etc.
- `gamma`: Influences how far the influence of a single point reaches (used in RBF/poly)

---

## 📊 7. Strengths

- Works well in **high dimensions**
- Effective with **clear margin of separation**
- Can handle **non-linear** classification with kernels
- Memory efficient (only support vectors are stored)

---

## ⚠️ 8. Limitations

- **Slow** on large datasets (complex training)
- **Sensitive to feature scaling** (always normalize!)
- Less interpretable than models like logistic regression
- Choosing the right **kernel and parameters** can be tricky

---

## Decision Tree Classifier

# 🌳 Understanding Decision Tree Classifier

A **Decision Tree Classifier** is a supervised learning model that splits the data into **branches** based on feature values, forming a tree-like structure to make predictions.

Each **internal node** represents a feature condition, each **leaf node** represents a class label, and each **branch** represents a decision rule.

---

## ✅ 1. Core Idea

- A Decision Tree repeatedly splits the dataset into subsets using **conditions** on features.
- It asks yes/no (or threshold-based) questions at each node to classify an input.
- The tree grows until it reaches:
  - Pure nodes (all samples belong to one class), or
  - A stopping condition (e.g., max depth, min samples per leaf)

---

## 🔍 2. Example of a Rule

Each **decision path** is an interpretable rule!

---

## 🧮 3. Splitting Criteria

To decide how to split nodes, the tree uses a **measure of impurity**:

- **Gini Impurity** (default in scikit-learn):
  \[
  Gini = 1 - \sum_{i=1}^{C} (p_i)^2
  \]

- **Entropy** (Information Gain):
  \[
  Entropy = - \sum_{i=1}^{C} p_i \log_2(p_i)
  \]

Where \( p_i \) is the proportion of class \( i \) at a node.

The algorithm chooses the **split that reduces impurity the most**.

---

## ⚙️ 4. Key Hyperparameters

- `criterion`: "gini" (default) or "entropy"
- `max_depth`: maximum depth of the tree
- `min_samples_split`: minimum samples required to split a node
- `min_samples_leaf`: minimum samples required at a leaf node
- `max_features`: number of features to consider when splitting

These help prevent **overfitting**, which decision trees are prone to.

---

## 📊 5. Strengths

- **Highly interpretable**
- Handles **both numerical and categorical** data
- **No need for feature scaling**
- Captures **non-linear relationships**
- Works well even with missing values (with proper handling)

---

## ⚠️ 6. Limitations

- Prone to **overfitting**, especially with deep trees
- Unstable (small changes in data can change the tree structure)
- **Not great with noisy data**
- Typically outperformed by ensembles like **Random Forest** or **XGBoost**

---
