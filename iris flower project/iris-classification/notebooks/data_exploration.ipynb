```python
# distributions for skew
# Sepal_length: the length of the leaf below the flower - continuous variable
# Sepal_width: the width of the leaf below the flower - continuous variable
# Petal_length: the length of the flower petal - continuous variable
# Petal_width: the width of the flower petal - continuous variable
# Species: the species of the flower - categorical variable

import matplotlib.pyplot as plt
import seaborn as sns

plt.hist(df['sepal_length'], bins=20, color='blue', alpha=0.7)
plt.hist(df['sepal_width'], bins=20, color='green', alpha=0.7)
plt.hist(df['petal_length'], bins=20, color='red', alpha=0.7)
plt.hist(df['petal_width'], bins=20, color='purple', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Distribution of Sepal and Petal Measurements')
plt.legend(['Sepal Length', 'Sepal Width', 'Petal Length', 'Petal Width'])
plt.show()

```

```python
# kde plot using sns
sns.kdeplot(data=df, x='sepal_length',fill=True, multiple="layer")
sns.kdeplot(data=df, x='sepal_width', fill=True, multiple="layer")
sns.kdeplot(data=df, x='petal_length',  fill=True, multiple="layer")
sns.kdeplot(data=df, x='petal_width',  fill=True, multiple="layer")
plt.legend(['Sepal Length', 'Sepal Width', 'Petal Length', 'Petal Width'])
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()

# are sepal lenght and width normally distributed?
"""
Shapiro-Wilk test for normality:
The test returns two statistics - the test statistic and the p-value.

Null Hypothesis (H0): The data is normally distributed.
Alternative Hypothesis (H1): The data is not normally distributed.

If p-value is <= 0.05, we reject the null hypothesis and conclude that the data is not normally distributed.
If p-value is > 0.05, we fail to reject the null hypothesis and conclude that the data is normally distributed.

The W statistic is a measure of how well the data fits a normal distribution. A W statistic close to 1 
indicates that the data is normally distributed, while a W statistic significantly less than 1 indicates that the data is not normally distributed.
"""
from scipy.stats import shapiro
stat, p = shapiro(df['sepal_length'])
statl, pl = shapiro(df['sepal_width'])

print(f'Sepal Length: Statistics={stat:.3f}, p-value={p:.3f}')
print(f'Sepal Width: Statistics={statl:.3f}, p-value={pl:.3f}')
```
